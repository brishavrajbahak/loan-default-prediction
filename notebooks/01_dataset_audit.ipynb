### Prediction Target Defination

The model predicts the final outcome of completed loans.

- `final_good` is encoded as `default_flag = 0`.
- `final_bad` is encoded as `default_flag = 1`.
- Current, late, or other unresolved status are not used as training targets.
- `outcome_group` is retained for business interpretation.
- `default_falg` is used as machine-learning target beacuse it provides consistent binary value.

The target validation returned zero unexpected rows, confirming that `outcome_group` and `default_flag` agree for every record.

```python
assert set(df["outcome_group"].dropna().unique()) == {
    "final_good",
    "final_bad",
}

assert set(df["default_flag"].dropna().unique()) == {0,1}

assert (
    ((df["outcome_group"] == "final_good") & (df["default_flag"] == 0)) |
    ((df["outcome_group"] == "final_bad") & (df["default_flag"] == 1))
).all()

print("All targets validation check passed.")
```

## Day 3 - Data Leakage Audit and PostgreSQL Setup

```python
allowed_features = [
    "loan_amnt",
    "term_months",
    "int_rate",
    "grade",
    "sub_grade",
    "emp_length_years",
    "home_ownership",
    "annual_inc",
    "verification_status",
    "issue_year",
    "purpose",
    "dti",
]

excluded_features = [
    "id",
    "issue_date",
    "issue_month",
    "loan_status",
    "outcome_group",
    "default_flag",
]

print("Alowed features:", len(allowed_features))
print(allowed_features)

print("\nExcluded_features:", len(excluded_features))
print(excluded_features)
```

```python
dataset_columns = set(df.columns)

missing_allowed = set(allowed_features) - dataset_columns
unknown_excluded = set(excluded_features) - dataset_columns
overlap = set(allowed_features) & set(excluded_features)

print("Allowed features missing from dataset:", missing_allowed)
print("Excluded feature not found in dataset:", unknown_excluded)
print("Features appearing in both lists:", overlap)

assert len(missing_allowed) == 0
assert len(overlap) == 0

print("Feature boundary check passed.")
```

```python
model_columns = allowed_features + ["default_flag"]

model_df = df[model_columns].copy()

model_df["default_flag"] = model_df["default_flag"].astype("int8")

print("Modeling dataset shape:", model_df.shape)
display(model_df.head())
```

```python
MODEL_DATA_PATH = "../data/processed/modeling_datset.csv"

model_df.to_csv(MODEL_DATA_PATH, index=False)

print(f"Saved modeling dataset to: {MODEL_DATA_PATH}")
```

```python
feature_columns = allowed_features
target_column = "default_flag"

feature_leakage = set(feature_columns) & set(excluded_features)
target_in_features = target_column in feature_columns

print("Feature columns:", feature_columns)
print("Feature leakage found:", feature_leakage)
print("Target included in features:", target_in_features)

assert feature_leakage == set()
assert target_in_features is False

print("Leakage audit passed.")
```

### Leakage Audit Summary

The model input features contain only information available at loan origination.

- `default_flag` is the prediction target, not an input feature.
- `loan_status` and `outcome_group` are excluded because they reveal the final outcome.
- `id` is excluded because it is an identifier rather than a risk characteristic.
- Payment, recovery, collection, and last-payment fields are excluded because they occur after origination.
- The leakage audit returned an empty set, so no excluded field appears in the input features.
- The approved modeling dataset contains 12 input features and 1 target column.
