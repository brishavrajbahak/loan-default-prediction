```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path("../data/processed/loan_default_final_outcomes.csv")

df = pd.read_csv(DATA_PATH)
df.shape
```

```python
print("Rows and coumns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()
```

```python
missing_summary = (
    df.isna()
    .sum()
    .to_frame("missing_count")
    .assign(
        missing_percentage = lambda table:
        (table["missing_count"] / len(df) * 100).round(2)
    )
    .sort_values("missing_count", ascending=False)
)

display(missing_summary)

print("Duplicate rows:", df.duplicated().sum())
print("Duplicated loan IDs:", df["id"].duplicated().sum())
```

```python
print("Outcome groups:")
display(df["outcome_group"].value_counts(dropna=False))

print("Default_flag:")
display(df["default_flag"].value_counts(dropna=False))

print("Default_rate:")
print(df["default_flag"].mean().round(4) * 100)
```

```python
display(
    pd.crosstab(
        df["outcome_group"],
        df["default_flag"],
        margins=True
    )
)
```
